## 1. What is a Conditional?

A conditional statement allows a program to make a decision.

It checks whether a condition is `True` or `False`.

The basic structure is:

if condition:
    do something

If the condition is True, Python executes the indented code.
If the condition is False, Python skips it.

```python
age = 20
if age >= 18:
    print("adult")
```

### Explanation

- `if` starts the conditional statement.
- `age >= 18` is the condition.
- `:` indicates that the conditional block starts.
- The indented `print()` statement runs if the condition is True.

## 2. Comparison Operators

Comparison operators compare two values.

| Operator | Meaning |
|----------|---------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

The result of a comparison is always a Boolean value:
`True` or `False`.


## =   → assignment
## ==  → comparison

```python
x = 10

print(x == 10)
print(x > 5)
print(x < 5)
print(x != 20)
```

## 3. `if` and `else`

Sometimes we want the program to do one thing when a condition is True
and another thing when the condition is False.

We can use `else` for this.

```python
reward = -1
if reward > 0:
    print ("Excellent")
else:
    print ("Bad reward")
```

## 4. `elif`

`elif` means "else if".

It allows us to check another condition if the previous condition was False.

```python
reward = 4 

if reward > 5:
    print ("high reawrd")
elif reward > 0:
    print ("positive reward")
else:
    print ("negative reward")
```

## 5. Logical Operators

We can combine conditions using logical operators.

| Operator | Meaning |
|----------|---------|
| `and` | Both conditions must be True |
| `or` | At least one condition must be True |
| `not` | Reverses True and False |

```python
age = 20
HS = has_permission = True

if age >= 18 and HS:
    print ("Access Allowed")
```

## 6. What is a Loop?

A loop allows us to repeat an action multiple times.

Instead of writing:

print(0)
print(1)
print(2)
print(3)

we can use a loop to repeat the operation.

## 7. `for` Loops

A `for` loop is used when we want to repeat an action for
each item in a sequence.

```python
for i in range(5):
    print(i)
```

```python
actions = ["left", "right" , "up" , "down"]

for action in actions:
    print (action)
```

```python
## for Loop + if

rewards = [10, -2, 5, -1, 20]

for reward in rewards:
    if reward > 0:
        print ("positive", reward)
```

### Using `enumerate()`

When we loop through a list, sometimes we need both:

- the value of the item
- the index (position) of the item

`enumerate()` gives us both.

For example:

`enumerate(rewards)` gives us pairs such as:

`(0, 10)`, `(1, -2)`, `(2, 5)`, ...

The first number is the index and the second number is the value.

```python
rewards = [10, -2, 5, -1, 20]

for index, reward in enumerate(rewards):
    if reward > 0:
        print(index, reward, "positive")
```

## 8. `while` Loops

A `while` loop repeats an action as long as a condition is True.

Unlike a `for` loop, we don't necessarily know in advance
how many times the loop will run.

```python
count = 0

while count < 5:
    print (count)
    count = count + 1
```

## 9. `for` vs `while`

A `for` loop is useful when we want to go through a sequence
or repeat something a known number of times.

A `while` loop is useful when we want to continue repeating
something as long as a condition remains True.

### Simple rule

`for` → "For each item / repeat this many times"

`while` → "Keep doing this while this condition is True"

## 10. Connection to Reinforcement Learning

Reinforcement Learning involves repeated decision-making.

At each step:

1. The agent observes a state.
2. The agent chooses an action.
3. The environment gives a reward.
4. The agent moves to a new state.
5. The process continues.

Python conditionals and loops can be used to represent this logic.

```python
state = 0

while state < 5:

    if state < 4:
        action = "move right"
    else:
        action = "reach goal"

    print("State:", state)
    print("Action:", action)

    state = state + 1
```
