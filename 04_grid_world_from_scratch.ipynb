```python
grid = [
    [".", ".", ".", "."],
    [".", ".", ".", "."],
    [".", ".", ".", "."],
    [".", ".", ".", "."]
]

agent_position = [3, 0]
goal_position = [0, 3]

actions = ["up", "down", "left", "right"]

action = "right"

row = agent_position[0]
column = agent_position[1]

if action == "up" and row > 0:
    agent_position[0] -= 1

elif action == "down" and row < 3:
    agent_position[0] += 1

elif action == "left" and column > 0:
    agent_position[1] -= 1

elif action == "right" and column < 3:
    agent_position[1] += 1

if agent_position == goal_position:
    reward = 10
else:
    reward = -1

grid[agent_position[0]][agent_position[1]] = "A"
grid[goal_position[0]][goal_position[1]] = "G"

for row in grid:
    print(" ".join(row))

print("Agent position:", agent_position)
print("Action:", action)
print("Reward:", reward)
```

```python
import matplotlib.pyplot as plt

def plot_grid(agent_position, goal_position):
    fig, ax = plt.subplots()

    # Draw the grid
    ax.set_xlim(0, 4)
    ax.set_ylim(0, 4)

    # Add grid lines
    ax.set_xticks(range(5))
    ax.set_yticks(range(5))
    ax.grid()

    # Plot the agent
    ax.scatter(agent_position[1] + 0.5,
               3 - agent_position[0] + 0.5,
               s=500,
               label="Agent")

    # Plot the goal
    ax.scatter(goal_position[1] + 0.5,
               3 - goal_position[0] + 0.5,
               s=500,
               label="Goal")

    ax.legend()
    plt.show()
```

```python
## Let's make a simple plotting function

import matplotlib.pyplot as plt

agent_position = [3, 0]
goal_position = [0, 3]

actions = ["right", "right", "up", "up", "up", "right"]

action_index = 0

while agent_position != goal_position:

    action = actions[action_index]

    row = agent_position[0]
    column = agent_position[1]

    if action == "up" and row > 0:
        agent_position[0] -= 1

    elif action == "down" and row < 3:
        agent_position[0] += 1

    elif action == "left" and column > 0:
        agent_position[1] -= 1

    elif action == "right" and column < 3:
        agent_position[1] += 1

    print("Action:", action)
    print("Position:", agent_position)

    plot_grid(agent_position, goal_position)

    action_index += 1

```
