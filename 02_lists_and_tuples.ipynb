## 1. What is a List?

A list is a collection of multiple values stored in a single variable.

For example, instead of creating three separate variables for
three possible actions, we can store them together in one list.

```python
##  practice

actions = ["left", "right", "stay"]

print(actions)
```

### Explanation

`actions` is the variable name.

The square brackets `[]` tell Python that we are creating a list.

The list contains three strings:

- `"left"`
- `"right"`
- `"stay"`

## 2. Accessing Items in a List

We can access individual items in a list using their index.

Python uses zero-based indexing.

This means that the first item has index `0`, not `1`.

```python
actions = ["left", "right", "stay"]

print(actions[0])
print(actions[1])
print(actions[2])

##  Python uses zero-based indexing, so the first element is at index 0.
```

## 3. Modifying a List

Lists are mutable.

This means that we can change their contents after creating them.

```python
actions = ["left", "right", "stay"]

actions[2] = "jump"

print(actions)
```

## 4. Adding Items to a List

We can add an item to the end of a list using `append()`.

```python
actions = ["left", "right"]

actions.append("stay") ## Add "stay" to the end of the actions list.

print(actions)
```

## 5. Removing Items from a List

We can remove an item using the `remove()` method.

```python
actions = ["left", "right", "stay"]

actions.remove("stay")

print(actions)
```

## 6. Length of a List

The `len()` function tells us how many items are in a list.

```python
actions = ["left", "right", "stay"]

print(len(actions))
```

## 7. Looping Through a List

A `for` loop allows us to process each item in a list one at a time.

```python
actions = ["left", "right", "stay"]

for action in actions:
    print(action)
```

```python
for i in range(10):
    print(i)
```

## 8. What is a Tuple?

A tuple is another collection of values.

A tuple looks similar to a list, but we use parentheses `()`.

The main difference is that tuples are immutable.

This means that once a tuple is created, we cannot change its elements.

```python
position = (10, 20)

print(position)
```

```python
position = [10, 20]

position[0] = 15

#List can be modified

print(position)
```

```python
position = (10, 20)

position[0] = 15

# you will get error because tuples cannot be modified.

print(position)
```

```python
##  Indexing Also Works With Tuples

position = (10, 20)

print(position[0])
print(position[1])
```

## Connection to Reinforcement Learning

Lists are useful when we have a collection of values that may change.

For example, an agent may have a collection of possible actions:

- left
- right
- up
- down

We can represent them using a list.

```python
##  A list can also store rewards collected during an episode.

rewards = [0, -1, -1, 10]

total_reward = sum(rewards)

print("Rewards:", rewards)
print("Total reward:", total_reward)
```
