## Checking the Data Type

Python provides the `type()` function to tell us what type of
value a variable contains.

```python
name = "Maryam"
age = 35
height = 1.65
is_student = True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

```python
## Explanation

"Maryam" → text → str

35 → whole number → int

1.65 → decimal → float

True → yes/no → bool
```

## Basic Arithmetic with Variables

Variables can be used in mathematical calculations.

Python supports common arithmetic operations such as:

- `+` addition
- `-` subtraction
- `*` multiplication
- `/` division

```python
reward = 10
penalty = 3

total = reward - penalty

print(total)
```

## Connection to Reinforcement Learning

In Reinforcement Learning, we constantly store information

```python
## example
state = 2
action = 1
reward = 10

print("State:", state)
print("Action:", action)
print("Reward:", reward)
```

# Python Data Types and Data Structures

Python has several built-in types for storing and organizing information.

## 1. Basic Data Types

These are the basic types we use to represent individual values:

| Type | Meaning | Example |
|------|---------|---------|
| `int` | Whole number | `25` |
| `float` | Decimal number | `3.14` |
| `str` | Text | `"Maryam"` |
| `bool` | True or False | `True` |

## 2. Data Structures

These types are used to store collections of values:

| Type | Meaning | Example |
|------|---------|---------|
| `list` | Ordered and changeable collection | `[1, 2, 3]` |
| `tuple` | Ordered and unchangeable collection | `(1, 2, 3)` |
| `dict` | Collection of key-value pairs | `{"name": "Maryam"}` |
| `set` | Collection of unique values | `{1, 2, 3}` |

### Accessing Items in a List

Each item in a list has an index.

Python starts counting from `0`.

For example:

- `actions[0]` → first item
- `actions[1]` → second item
- `actions[2]` → third item
- `actions[3]` → fourth item

```python
actions = ["left", "right", "up", "down"]

print(actions)
print(actions[0])
print(actions[1])
print(actions[2])
print(actions[3])
```
